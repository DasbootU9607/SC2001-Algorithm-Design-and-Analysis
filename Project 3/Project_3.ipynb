### Project 3

Give a dynamic programming algorithm to compute the maximum profit, given a 
knapsack of capacity C, n types of objects with weights $w_i$ and profits $p_i$ using the 
bottom up approach.


### Bottom-Up Dynamic Programming Algorigthm

```text
Algorithm dp_knapsack(C, n, w, p)

Input:
    C: Knapsack capacity
    n: Number of item types
    w: Array of item weights
    p: Array of item profits

Output:
    Maximum total profit P(C)

    P = [0] * (C + 1)

    for c from 1 to C:
        for i from 0 to n - 1:
            if w[i] <= c:
                P[c] = max(P[c], P[c-w[i]] + p[i])

    return P[C]
```

#### Code implementation

````python
def knapsack(C, weights, profits):
    n = len(weights)
    # Initialize dp array
    P = [0] * (C + 1)

    # Traverse from 1 to bottom up
    for c in range(1, C + 1):
        for i in range(n):
            if weights[i] <= c:
                P[c] = max(P[c], P[c-weights[i]] + profits[i])
    
    return P[C]
````

````python
# Task 4(a)
weights1 = [4, 5, 8]
profits1 = [7, 6, 9]
C1 = 16

print("Task 4(a): P(16) =", knapsack(C1, weights1, profits1))
````

````python
# Task 4(b)
weights2 = [5, 6, 7]
profits2 = [7, 6, 9]
C2 = 12

print("Task 4(b): P(12) =", knapsack(C2, weights2, profits2))
````
